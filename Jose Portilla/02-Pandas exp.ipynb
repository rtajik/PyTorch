# Different .apply .map and .applymap

## Source - https://stackoverflow.com/a/19798528
## Posted by jeremiahbuddha, modified by community. See post 'Timeline' for change history
## Retrieved 2026-08-26, License - CC BY-SA 4.0

```python
frame = pd.DataFrame(np.random.randn(4, 3), columns=list('bde'), index=['Utah', 'Ohio', 'Texas', 'Oregon'])
```

```python
f = lambda x: x.max() - x.min()

frame.apply(f)
```

```python
format = lambda x: '%.2f' % x
frame.applymap(format)
```

```python
frame['e'].map(format)
```
