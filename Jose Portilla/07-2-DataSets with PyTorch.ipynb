```python
import torch
import torch.nn as nn

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
iris = pd.read_csv('Data/iris.csv')
```

```python
iris.head()
```

```python
iris.tail()
```

```python
Features = iris.drop('target', axis=1).values
Features = torch.FloatTensor(Features)
```

```python
# Labels = iris['target'].values.reshape(-1, 1)
Labels = iris['target'].values
Labels = torch.LongTensor(Labels)
```

```python
print(Features.type(), Labels.type())
```

```python
from torch.utils.data import TensorDataset, DataLoader
```

```python
iris_torchDataset = TensorDataset(Features, Labels)
```

```python
for X, y in iris_torchDataset:
    print(X.shape, y.shape, sep = '\n')
```

```python
iris = DataLoader(iris_torchDataset, batch_size=10, shuffle=True)
for i, Data in enumerate(iris):
    print(i, Data)
```
