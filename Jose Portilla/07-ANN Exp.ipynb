```python
import torch
```

```python
x1 = torch.tensor(2.0, requires_grad=True)
x2 = torch.tensor([[1., 2.0, 3.],[3.,2.,1.]], requires_grad=True)
```

```python
# y = 2*x**4 + x**3 + 3*x**2 + 5*x + 1
# y = 2*x**4
y1 = 3*x1**2 + 1
y2 = 3*x2**2 + 1
```

```python
print(x1,y1, sep='\n')
```

```python
y1.backward()
print(x1.grad)
```

```python
print(x2,y2, sep='\n')
```
